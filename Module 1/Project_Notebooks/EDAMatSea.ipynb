# ***`Project: Employee Salary Analysis using Matplotlib & Seaborn`***

Objective: Analyze employee salary patterns based on experience, education, job role, location, and skills using Matplotlib and Seaborn.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("/content/salary_dataset (1) (1).csv")

df.head()
```

```python
#Univariate Analysis
#Distribution by salary
plt.figure(figsize=(8,5))

sns.histplot(df["Salary"], bins=10, kde=True)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.show()
```

```python
#Experience Distribution
plt.figure(figsize=(8,5))

sns.histplot(df["Experience"], bins=10, kde=True)

plt.title("Experience Distribution")
plt.xlabel("Years of Experience")
plt.ylabel("Number of Employees")
plt.show()
```

```python
#Number of employees by jobrole
plt.figure(figsize=(10,5))

sns.countplot(data=df, x="JobRole")

plt.title("Number of Employees by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)

plt.show()
```

```python
#Salary by JobRole
#Bi-variate analysis(including two Variables)
plt.figure(figsize=(10,5))

sns.barplot(data=df, x="JobRole", y="Salary")

plt.title("Average Salary by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)

plt.show()
```

```python
#Salary by Education
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Education", y="Salary")

plt.title("Average Salary by Education")
plt.xlabel("Education")
plt.ylabel("Average Salary")

plt.show()
```

```python
#salary by location
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Location", y="Salary")

plt.title("Average Salary by Location")
plt.xlabel("Location")
plt.ylabel("Average Salary")

plt.show()
```

```python
#Experience Vs Salary
plt.figure(figsize=(8,5))

sns.scatterplot(data=df, x="Experience", y="Salary")

plt.title("Experience vs Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")

plt.show()
```

```python
#With Regression line
plt.figure(figsize=(8,5))

sns.regplot(data=df, x="Experience", y="Salary")

plt.title("Experience vs Salary with Trend Line")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")

plt.show()
```

```python
#Salary Distribution with Eduction
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x="Education", y="Salary")

plt.title("Salary Distribution by Education")
plt.xlabel("Education")
plt.ylabel("Salary")

plt.show()
```

```python
#Salary distribution by jobRole
plt.figure(figsize=(10,5))

sns.boxplot(data=df, x="JobRole", y="Salary")

plt.title("Salary Distribution by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Salary")
plt.xticks(rotation=45)

plt.show()
```

```python
#Location Vs Job Role
plt.figure(figsize=(10,5))

sns.countplot(data=df, x="Location", hue="JobRole")

plt.title("Job Roles Across Locations")
plt.xlabel("Location")
plt.ylabel("Number of Employees")

plt.show()
```

```python
#Correlation Heatmap
#collect numeric columns
numeric_df = df[["Experience", "Salary"]]
```

```python
plt.figure(figsize=(6,4))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
#Combined Visualisation
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="Experience",
    y="Salary",
    hue="Education",
    style="Education",
    s=100
)

plt.title("Experience vs Salary by Education")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")

plt.show()
```

## Answers to Questions

```python
# 1. What is the average salary?
average_salary = df['Salary'].mean()
print(f"The average salary is: ${average_salary:,.2f}")
```

```python
# 2. What is the maximum salary?
maximum_salary = df['Salary'].max()
print(f"The maximum salary is: ${maximum_salary:,.2f}")
```

```python
# 3. Which job role has the highest average salary?
job_role_avg_salary = df.groupby('JobRole')['Salary'].mean().sort_values(ascending=False)
highest_avg_job_role = job_role_avg_salary.index[0]
highest_avg_job_role_salary = job_role_avg_salary.iloc[0]
print(f"The job role with the highest average salary is '{highest_avg_job_role}' with an average salary of: ${highest_avg_job_role_salary:,.2f}")
```

```python
# 4. Which location has the highest average salary?
location_avg_salary = df.groupby('Location')['Salary'].mean().sort_values(ascending=False)
highest_avg_location = location_avg_salary.index[0]
highest_avg_location_salary = location_avg_salary.iloc[0]
print(f"The location with the highest average salary is '{highest_avg_location}' with an average salary of: ${highest_avg_location_salary:,.2f}")
```

## 5. Does salary increase with experience?
Yes, as observed in the 'Experience vs Salary with Trend Line' plot, there is a clear positive linear relationship between years of experience and salary. As experience increases, salary generally tends to increase.

```python
# 6. Which education level has the highest average salary?
education_avg_salary = df.groupby('Education')['Salary'].mean().sort_values(ascending=False)
highest_avg_education = education_avg_salary.index[0]
highest_avg_education_salary = education_avg_salary.iloc[0]
print(f"The education level with the highest average salary is '{highest_avg_education}' with an average salary of: ${highest_avg_education_salary:,.2f}")
```

```python
# 7. Which job role has the widest salary distribution?
job_role_salary_std = df.groupby('JobRole')['Salary'].std().sort_values(ascending=False)
widest_dist_job_role = job_role_salary_std.index[0]
widest_dist_std = job_role_salary_std.iloc[0]
print(f"The job role with the widest salary distribution (highest standard deviation) is '{widest_dist_job_role}' with a standard deviation of: ${widest_dist_std:,.2f}")
```

## 8. Are there any salary outliers?
Yes, by examining the 'Salary Distribution by Job Role' box plot, we can observe several outliers (represented by individual points outside the whiskers) for various job roles, such as 'UX Designer', 'Data Scientist', 'HR Manager', and 'Software Engineer'.

```python
# 9. Which location has the most employees?
location_employee_count = df['Location'].value_counts()
most_employees_location = location_employee_count.index[0]
most_employees_count = location_employee_count.iloc[0]
print(f"The location with the most employees is '{most_employees_location}' with {most_employees_count} employees.")
```

## 10. What relationship do you observe between Experience and Salary?
There is a strong positive linear relationship between Experience and Salary. As an employee's years of experience increase, their salary tends to increase proportionally. This trend is clearly visible in both the scatter plot and the regression plot of 'Experience vs Salary'.
